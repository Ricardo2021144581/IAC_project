Question 7.1
In at most half a page, state which decisions of the pipeline changed the result most, and which did
not change it.

Decisions that changed the result most:

-Evaluation protocol. The same SVM pipeline scored a macro F1 of 0.761 with a random 70/30 split but only 0.568 under leave-one-subject-out (LOSO). The random split places videos of the same actor in both training and test, so the model is partly tested on faces it has already seen. LOSO tests on a person never seen in training, which is what a real user is. The 0.19 gap is the largest effect in the project, and it shows that the random split overestimates the performance on new people.

-Value of C. With the 55 features and LOSO, the macro F1 was 0.395 for C = 0.01 and 0.425 for C = 0.1, but 0.568 for C = 1. A very small C regularises the SVM too strongly, so it underfits. For C ≥ 1 the score stays at about 0.55–0.57.

-Reducing the number of features. Compared with using all 55 features (0.568), PCA keeping only 75% of the variance (9 components) lowered the macro F1 to 0.510, a drop of about 0.06. Selecting 10 features also gave clearly lower scores: 0.416 with the ANOVA F-statistic and 0.481 with mRMR. mRMR was better than ANOVA at the same K, probably because it avoids choosing features that carry the same information, like the highly correlated pairs found in 3.4. But neither selection beat keeping all the features.


Decisions that did not change it much:

-PCA with 95% of the variance. It reduces the 55 features to 24 components and scored 0.550, against 0.568 with all 55 features. The difference of 0.018 is small compared with the variation between actors (a standard deviation of about 0.12), so we cannot say that the compression made the classifier worse or better. It gives a smaller representation at a similar score.

-The extra features of 3.2. Using all 55 features (the original 30 plus the five new measures and their statistics) gave a macro F1 of 0.555, against 0.526 with only the original 30 features, both with C chosen by nested cross-validation. The gain of 0.029 is better for 14 of the 20 actors, but it is small compared with the variation between actors, so the evidence is suggestive and not conclusive.

-The number of features in mRMR. Going from 10 to 15 features raised the macro F1 only from 0.481 to 0.486, so extra features beyond 10 added almost nothing.

-Grouping the inner folds by actor. In the nested procedure, using StratifiedKFold (which ignores the actor) instead of GroupKFold did not change the outer macro F1 (0.554 against 0.555) or much the chosen C. It did change the inner estimate from 0.575 to 0.878, because the inner folds then contained videos of the same actors. This decision therefore did not change the result, but it changed how much the validation score could be trusted.

Question 7.2
The data were recorded by 24 professional actors, portraying emotions on request while reading two statements, facing one camera. Which of these conditions limit what the LOSO and test scores say about the users of a real application?

-Acted emotions. The actors portrayed emotions on request, so the expressions are deliberate and prototypical. People using the application would show emotions that are spontaneous, subtler, mixed or absent. Scores on acted data therefore likely overestimate how well the model recognises real emotions. The confusions between similar expressions (Surprised and Fearful, Calm and Neutral) would probably be worse with subtler expressions.

-Professional actors and a small sample. Professional actors are trained to be expressive and consistent, which most users are not. There are only 24 actors in total (20 for LOSO and 4 for the test), with little variation in age and appearance, so the model has seen a limited range of faces. The scores also vary a lot between actors: the standard deviation of the macro F1 in 4.2 was about 0.12, so the mean over the four test actors has a standard error of about 0.06. The test score (0.540) is therefore uncertain by more than its small difference from LOSO (0.550).

-Only two statements. All videos contain one of two sentences, so the mouth movements come from the same words. The model may partly use the movement of the mouth caused by speech, and not the emotion. A real user would say something different, or nothing, and the mouth landmarks would behave differently.

-One camera and one tracker. All videos were recorded in the same conditions, tracked with OpenFace, with frontal faces and controlled lighting. In a real application, the person uses their own camera, light and distance, and the tracker is MediaPipe. Our own recordings in 6.3 show this effect: many features fell outside the training range, and the number varied a lot between the two people (between 0 and 19 features per clip), even with the same tracker. The model’s behaviour on such inputs is not covered by the LOSO or test scores.
