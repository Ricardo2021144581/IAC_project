Question 7.1
In at most half a page, state which decisions of the pipeline changed the result most, and which did
not change it.

Decisions that most significantly altered the results:
    -Evaluation protocol used: With a random split, we obtained a macro-F1 score of 0.761, whereas using LOSO (Leave-One-Subject-Out), the macro-F1 score was only 0.569. This difference can be explained by the fact that the random split uses data from the same subject in both the training and testing sets, leading to data leakage.

    -Feature selection method: With k=10, mRMR yielded a score of 0.481, while ANOVA yielded 0.416. This difference occurs because ANOVA may select redundant features, whereas mRMR aims to select features that provide distinct information.

Decisions that did not significantly affect the results:
    -PCA: Retaining 95% of the variance did not yield significant improvements; using 24 components resulted in a score of 0.55, compared to 0.568 for the full feature set—a marginal improvement considered insignificant. Furthermore, when using PCA to retain 75% of the variance, the macro-F1 score decreased by 0.06, indicating that this change provided no benefit to the model.

    -Value of 'k' in mRMR: Testing values ​​of 'k' greater than 10 did not yield significant improvements; for instance, increasing 'k' from 10 to 15 resulted in an improvement of only approximately 0.005.

Question 7.2
The data were recorded by 24 professional actors, portraying emotions on request while reading two statements, facing one camera. Which of these conditions limit what the LOSO and test scores say about the users of a real application?

-Limited number of actors: in a real-world application, one would encounter a wide range of individual differences (age, appearance, etc.), yet this model was trained on only 24 actors, which raises questions regarding its generalization capability.

-The model was trained using only two utterances; however, in a real-world scenario, the movements of mouth landmarks can vary significantly, even for the same emotion.

-Finally, in a real-world application, individuals would record using their own equipment and within their own specific environments. Given that these conditions may differ substantially from those under which the model was trained, the actual results might deviate from expectations.
